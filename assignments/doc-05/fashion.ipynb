## Librerías necesarias

```python
import tensorflow as tf
from tensorflow import keras
import matplotlib.pyplot as plt
```

## Dataset

Carga de dataset Fashion MNIST como solicita el enunciado. Los datos son divididos en una parte de entrenamiento, y la otra parte de prueba.

```python
fashion_mnist = keras.datasets.fashion_mnist
(X_train, Y_train), (X_test, Y_test) = fashion_mnist.load_data()
```

Es necesario normalizar los datos extraídos del dataset, los cuales son valores de 0 a 255, los convertimos a valores de 0 a 1.

```python
X_train = X_train / 255.0
X_test = X_test / 255.0
```

## Modelo

Con lo anterior, ya es posible crear y compilar el modelo para su posterior entrenamiento.

```python
model = keras.Sequential([
    keras.layers.Flatten(input_shape=(28,28)),
    keras.layers.Dense(128, activation='relu'),
    keras.layers.Dense(10, activation='softmax')
])

model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

Las imagenes 28x28 pasan a ser vectores de 784; se define una capa oculta de 128 neuronas, y se determina la salida del modelo con 10 categorias/clases.

## Entrenamiento

```python
model.fit(X_train, Y_train, epochs=10)
```

## Evaluación

Podemos determinar la precisión del modelo.

```python
test_loss, test_acc = model.evaluate(X_test, Y_test)
print("Precision:", test_acc)
```

## Predicción

Aplicamos el modelo, para obtener las probabilidades de cada clase, y se puede comparar con el valor real.

```python
predictions = model.predict(X_test)

print("Prediccion:", predictions[0].argmax())
print("Real:", Y_test[0])
```
