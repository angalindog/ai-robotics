```python
# ══════════════════════════════════════════════════════════════════════════════
# EJERCICIO 5 — Algoritmo Genético para Evolución de Palabras con Síntesis de Voz
# ══════════════════════════════════════════════════════════════════════════════
# Instalar dependencias (ejecutar primero en Colab):
!pip install gtts wordfreq

from gtts import gTTS
from IPython.display import Audio, display
from wordfreq import top_n_list   # ← corrección: top_n_list retorna lista de palabras sueltas
import difflib
import random
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

# ── Inicialización de variables ───────────────────────────────────────────────
palabra_objetivo = "anticonstitucionalmente"
n_poblacion      = 77
prob_mutacion    = 0.59    # ← corrección: debe ser alta (0.5) para evitar estancamiento
generaciones     = 2000

# Banco de palabras reales en español (top 10000 palabras más frecuentes)
# top_n_list retorna lista de strings individuales, sin frases
VOCABULARIO = [p for p in top_n_list('es', 10000) if p.isalpha() and p != palabra_objetivo]

# ── Funciones de voz ──────────────────────────────────────────────────────────
def voz(poblacion, nombre_archivo="voz.mp3"):
    """Sintetiza toda la población y la reproduce. Equivale al parlante del computador."""
    texto = " ".join(poblacion)
    tts   = gTTS(text=texto, lang="es")
    tts.save(nombre_archivo)
    display(Audio(nombre_archivo))

def voz_palabra(palabra, nombre_archivo="palabra.mp3"):
    """Sintetiza y reproduce una sola palabra."""
    tts = gTTS(text=palabra, lang="es")
    tts.save(nombre_archivo)
    display(Audio(nombre_archivo))

# ── Generación de la población inicial ───────────────────────────────────────
def pob_init():
    """50 palabras reales en español, ninguna igual al objetivo."""
    return random.sample(VOCABULARIO, n_poblacion)

# ── Función de aptitud ────────────────────────────────────────────────────────
def aptitud(palabra, objetivo):
    """
    Similitud entre la palabra candidata y el objetivo.
    difflib.SequenceMatcher retorna un ratio en [0, 1].
    1.0 = palabras idénticas.
    """
    return difflib.SequenceMatcher(None, palabra, objetivo).ratio()

# ── Selección por ruleta proporcional a aptitud ───────────────────────────────
def seleccion(palabras):
    pesos = [max(aptitud(p, palabra_objetivo), 0.001) for p in palabras]
    return random.choices(palabras, weights=pesos, k=2)

# ── Cruce por punto medio ─────────────────────────────────────────────────────
def cruce(padre1, padre2):
    # Si las longitudes son distintas, elegimos un punto de corte aleatorio basado en la más corta
    punto = random.randint(1, min(len(padre1), len(padre2)))
    hijo = padre1[:punto] + padre2[punto:]
    return hijo

# ── Mutación: reemplazar por otra palabra aleatoria ───────────────────────────
# ── Nueva Mutación: Ajuste fino de caracteres ───────────────────────────
def mutacion(prob_mut, palabra):
    if random.random() <= prob_mut:
        letras = "abcdefghijklmnopqrstuvwxyz"
        lista_palabra = list(palabra)
        indice = random.randint(0, len(lista_palabra) - 1)

        tipo = random.random()
        if tipo < 0.33: # Cambiar una letra
            lista_palabra[indice] = random.choice(letras)
        elif tipo < 0.66 and len(lista_palabra) > 1: # Borrar una letra
            lista_palabra.pop(indice)
        else: # Añadir una letra
            lista_palabra.insert(indice, random.choice(letras))

        return "".join(lista_palabra)
    return palabra

# ══════════════════════════════════════════════════════════════════════════════
# EJECUCIÓN PRINCIPAL
# ══════════════════════════════════════════════════════════════════════════════
historial_mejor    = []
historial_promedio = []

print("=" * 55)
print("  EJERCICIO 5 — Evolución de Palabras con Síntesis de Voz")
print("=" * 55)
print(f"  Palabra objetivo : '{palabra_objetivo}'")
print(f"  Población        : {n_poblacion} palabras")
print(f"  Generaciones máx : {generaciones}")
print(f"  P(mutación)      : {prob_mutacion}")
print("=" * 55)

print("\n🔊 Población Inicial — escucha 50 palabras aleatorias:")
poblacion = pob_init()
print(f"  Muestra: {poblacion[:8]} ...")
voz(poblacion, "poblacion_inicial.mp3")

print(f"\nPalabra objetivo: '{palabra_objetivo}' — escúchala aquí:")
voz_palabra(palabra_objetivo, "objetivo.mp3")
print()

for i in range(generaciones):

    poblacion = sorted(poblacion,
                       key=lambda x: aptitud(x, palabra_objetivo),
                       reverse=True)
    mejor     = poblacion[0]
    apt_mejor = aptitud(mejor, palabra_objetivo)
    apt_prom  = sum(aptitud(p, palabra_objetivo) for p in poblacion) / n_poblacion

    historial_mejor.append(apt_mejor)
    historial_promedio.append(apt_prom)

    print(f"  Generación {i:3d} | mejor: '{mejor:<10s}' | similitud: {apt_mejor*100:6.2f}%")

    # Audio en generaciones clave — se escucha cómo la palabra mejora
    if i in [0, 5, 10, 20, 30, 50]:
        print(f"  🔊 Gen {i}: '{mejor}'")
        voz_palabra(mejor, f"gen_{i:03d}_{mejor}.mp3")

    if apt_mejor == 1.0:
        print(f"\n Palabra objetivo encontrada en generación {i}!")
        break

    # Operadores genéticos
    nueva_poblacion = list(poblacion[:5])   # elitismo: top 5 sobreviven
    while len(nueva_poblacion) < n_poblacion:
        padres = seleccion(poblacion)
        hijo   = cruce(padres[0], padres[1])
        hijo   = mutacion(prob_mutacion, hijo)
        nueva_poblacion.append(hijo)
    poblacion = nueva_poblacion

# ── Resultado final ───────────────────────────────────────────────────────────
mejor_final     = sorted(poblacion, key=lambda x: aptitud(x, palabra_objetivo), reverse=True)[0]
similitud_final = aptitud(mejor_final, palabra_objetivo) * 100

print(f"\n Población Final — escucha cómo evolucionó:")
voz(poblacion, "poblacion_final.mp3")

print(f"\n Mejor resultado: '{mejor_final}'")
voz_palabra(mejor_final, "resultado_final.mp3")

# ── Gráficas ──────────────────────────────────────────────────────────────────
gens = len(historial_mejor)
gen_x = range(gens)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle(
    f"Ejercicio 5 — AG Evolución de Palabras con Síntesis de Voz\n"
    f"Objetivo: '{palabra_objetivo}'  |  Resultado: '{mejor_final}'  |  "
    f"Similitud: {similitud_final:.1f}%  |  Generación: {gens-1}",
    fontsize=12, fontweight='bold'
)

# Panel izquierdo: comparación carácter a carácter
ax    = axes[0]
largo = max(len(mejor_final), len(palabra_objetivo))
c_enc = mejor_final.ljust(largo)[:largo]
c_obj = palabra_objetivo.ljust(largo)[:largo]
colores = ['#2ecc71' if c_enc[j] == c_obj[j] else '#e74c3c' for j in range(largo)]

for j, ch in enumerate(c_obj):
    ax.text(j+0.5, 1.65, ch.upper(), ha='center', va='center',
            fontsize=26, fontweight='bold', color='#555555', fontfamily='monospace')

for j, (ch, col) in enumerate(zip(c_enc, colores)):
    ax.add_patch(mpatches.FancyBboxPatch((j+0.06, 0.55), 0.88, 0.90,
        boxstyle="round,pad=0.05", facecolor=col, alpha=0.22, edgecolor=col, linewidth=2))
    ax.text(j+0.5, 1.0, ch.upper(), ha='center', va='center',
            fontsize=26, fontweight='bold', color=col, fontfamily='monospace')

ax.text(-0.3, 1.65, 'Objetivo:',  ha='right', va='center', fontsize=10, color='#555555', style='italic')
ax.text(-0.3, 1.0,  'Resultado:', ha='right', va='center', fontsize=10, color='#333333', style='italic')
coincidencias = sum(1 for a, b in zip(c_enc, c_obj) if a == b)
ax.text(largo/2, 0.18, f'{coincidencias}/{largo} caracteres correctos  |  Similitud: {similitud_final:.2f}%',
        ha='center', va='center', fontsize=10, color='#333333')
ax.set_xlim(-0.5, largo+0.3); ax.set_ylim(-0.05, 2.1); ax.axis('off')
ax.set_title('Comparación carácter a carácter', pad=10)
ax.legend(handles=[mpatches.Patch(color='#2ecc71', alpha=0.5, label='Correcto'),
                   mpatches.Patch(color='#e74c3c', alpha=0.5, label='Incorrecto')],
          loc='lower right', fontsize=9)

# Panel derecho: convergencia
axes[1].plot(gen_x, [a*100 for a in historial_mejor],   color='#2ecc71', linewidth=2.2, label='Mejor (%)')
axes[1].plot(gen_x, [a*100 for a in historial_promedio], color='#3498db', linewidth=1.4, linestyle='--', alpha=0.8, label='Promedio (%)')
axes[1].fill_between(gen_x, [a*100 for a in historial_promedio], [a*100 for a in historial_mejor], alpha=0.15, color='#2ecc71')
axes[1].axhline(100, color='#e74c3c', linewidth=0.9, linestyle=':', alpha=0.7, label='100% = objetivo')
for g in [0, 5, 10, 20, 30, 50]:
    if g < gens:
        axes[1].axvline(g, color='#f39c12', linewidth=1.2, linestyle='--', alpha=0.55)
        axes[1].text(g+0.3, 3, 'Audio', fontsize=7, color='#f39c12', rotation=90)
axes[1].set_xlabel('Generación'); axes[1].set_ylabel('Similitud (%)')
axes[1].set_title(f"Convergencia — buscando '{palabra_objetivo}'")
axes[1].legend(fontsize=9); axes[1].grid(True, alpha=0.3); axes[1].set_ylim(0, 108)

plt.tight_layout()
plt.savefig("ejercicio5.png", dpi=120, bbox_inches='tight')
plt.show()

print("\n" + "="*55)
print(f"  Objetivo      : '{palabra_objetivo}'")
print(f"  Mejor hallada : '{mejor_final}'")
print(f"  Similitud     : {similitud_final:.2f}%")
print(f"  Generaciones  : {gens}")
```
